```python
from seispy.synthetics import synlv
from seispy.filters import butterworth_bandpass
from seispy.plotting import wiggle
import matplotlib.pyplot as plt
import numpy as np


segy = synlv().to_memory()
wiggle(segy)
plt.show()

bandpassed = butterworth_bandpass(synlv())

im_dat = np.array([trace for trace in bandpassed])

plt.figure()
plt.imshow(im_dat.T, clim=[-1, 1], cmap='seismic')
plt.show()
```

```python
def f(n):
    for i in range(n):
        print("here f")
        yield i

def g(src):
    for item in src:
        print("here g")
        yield 2 * item
```

```python
v = f(20)
gv = g(v)

for v in gv:
    print(v)
```
